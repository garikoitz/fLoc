```python
import os
import shutil
import random

basedir='/Users/tiger/toolboxes/fLoc/stimuli'

# Define input folders
body_folder = os.path.join(basedir,"body")
limb_folder = os.path.join(basedir,"limb")

# Define output folders
bodylimb1_folder = os.path.join(basedir,"bodylimb1")
bodylimb2_folder = os.path.join(basedir,"bodylimb2")

# Ensure output directories exist
os.makedirs(bodylimb1_folder, exist_ok=True)
os.makedirs(bodylimb2_folder, exist_ok=True)

# Get list of images
body_images = sorted([f for f in os.listdir(body_folder) if f.endswith('.jpg')])
limb_images = sorted([f for f in os.listdir(limb_folder) if f.endswith('.jpg')])

# Randomly select 80 images from each
random.shuffle(body_images)
random.shuffle(limb_images)

selected_body = body_images[:80]
selected_limb = limb_images[:80]

# Split into two sets
bodylimb1_list = selected_body[:40] + selected_limb[:40]
bodylimb2_list = selected_body[40:] + selected_limb[40:]

# Shuffle images within bodylimb1 and bodylimb2
random.shuffle(bodylimb1_list)
random.shuffle(bodylimb2_list)

# Ensure output directories exist
os.makedirs(bodylimb1_folder, exist_ok=True)
os.makedirs(bodylimb2_folder, exist_ok=True)

# Function to copy and rename images
def copy_and_rename(image_list, body_folder, limb_folder, dest_folder, prefix):
    for idx, img in enumerate(image_list, start=1):
        print(idx)
        print(img)
        if "body" in img:
            src_folder=body_folder
        elif "limb" in img:
            src_folder=limb_folder

        print(f"this is src_folder {src_folder}")
        src_path = os.path.join(src_folder, img)
        dest_path = os.path.join(dest_folder, f"{prefix}-{idx}.jpg")
        shutil.copy(src_path, dest_path)

# Copy and rename images
copy_and_rename(bodylimb1_list, body_folder , limb_folder, bodylimb1_folder, "bodylimb1")
copy_and_rename(bodylimb2_list, body_folder, limb_folder, bodylimb2_folder, "bodylimb2")

print("Images successfully shuffled, split, and renamed in 'bodylimb1' and 'bodylimb2' folders.")
```
